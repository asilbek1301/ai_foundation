```python
import numpy as np

# 1D array

arr = np.array([1, 3, 5, 7])
print('numpy_array:', arr)

arr.shape
```

```python
# 2D array Matrix

arr2 = np.array([[1, 3, 5, 7], [4, 6, 8, 9]])
print('numpy_array2:', arr2)

arr2.shape
```

```python
# dtype()  -->  arraydagi elementlarni type ni aniqlab beradi
a = np.array([1, 2, 3])
print(a.dtype)
```

```python
# reshape()  -->  yaratilgan arrayni shape ni o'zgartiradi
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)
print("shaped_array:", b)
```

```python
# astype()  -->  datani type ni o'zgartiradi
a = np.array([1, 2, 3])
b = a.astype(float)
print("float:", b)
```

```python
# np.zeros(), np.ones()  -->  arraylarni yaratadi
z = np.zeros((2, 3))
o = np.ones((2, 3))

print("zeros:", z)
print("ones:", o)
```

```python
# ndim -->  arrayda nechta Dimension(o'lchov) borligini aniqlaydi
a = np.array([[1, 2], [3, 4]])
print("ndim:", a.ndim)
```

```python
# arrange()  -->
a = np.arange(1, 7)
print("arange:", a)
```

```python
# sum(), mean(), max(), min()  --> hisob-kitob qilishda

a = np.array([1, 2, 3, 4,])

print("sum", a.sum())
print("mean", a.mean()) # o'rtacha
print("max", a.max())
print("min", a.min())
```

```python
# flatten()  -->  2D or 3D arrayni 1D listga aylantiradi

a = np.array([[1, 2], [3, 4]])
print("flatten:", a.flatten())
```

```python
# T (Transpose)
a= np.array([[1, 2], [3, 4]])
print("transpose:", a.T)
```

```python
# dot() @  -->  matrix larni ko'paytirayotganda
a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 2]])
print("dot/multiplication:", a.dot(b))
```

```python
# random.rand()  -->  array hosil qilganda random raqamlar hosil qiladi
a = np.random.rand(2, 3) # => 0.0 and 1.0 / 0.0 included / 1.0 excluded
print("random rand:", a)
```
